## Extract years from dates

```python
import duckdb

# First check
duckdb.sql("""--sql
SELECT Calendar_Date, SUBSTRING(Calendar_Date, 1, 5) AS test
FROM df
LIMIT 5;
""").df()

# Noticed became wrong when year was shorter than 4 digits or missing -
```

```python
# Create Year column and extract year, also remove year from Date column

df["Year"] = df["Calendar_Date"].str.extract(r"^(-?\d{1,4})", expand=False)
df["Year"] = pd.to_numeric(df["Year"])
df["Calendar_Date"] = df["Calendar_Date"].str.replace(r"^(-?\d{1,4})", "", regex=True)
df.head(2)
```

```python
df["Year"].value_counts().head(5)
```

## Create column Centuries

```python
# creation
import numpy as np

df["Century"] = np.where(
    df["Year"] > 0,
    np.ceil(df["Year"] / 100),
    np.floor(df["Year"] / 100)
).astype(int)

df.head(3)
        
```

```python
def century(century):
    century_str = str(century)
    if century_str in ["11", "12", "13"]:
        suffix = "th"    
    elif century_str[-1] == "1":
        suffix = "st"
    elif century_str[-1] == "2":
        suffix = "nd"
    elif century_str[-1] == "3":
        suffix = "rd"
    else:
        suffix = "th"

    if century < 0:
        return f"{century_str}{suffix} century BCE"
    else:
        return f"{century_str}{suffix} century"

df["Century"] = df["Century"].apply(century)
```

## Eclipse durations

```python
# replace - with 0

def convert_time(column):
    df[column] = df[column].str.replace("-", "0")
    df[column] = pd.to_numeric(df[column], errors="coerce").fillna(0)
```

```python
convert_time("Partial_Eclipse_Duration_(m)")
convert_time("Total_Eclipse_Duration_(m)")

df.head(2)
```

```python
# Calculate total time for eclipse

df["Eclipse_Total_Time"] = df["Penumbral_Eclipse_Duration_(m)"] + df["Partial_Eclipse_Duration_(m)"] + df["Total_Eclipse_Duration_(m)"]
df.head(2)
```

## Replace eclipse type with name of type

```python
df["Eclipse_Group_Type"] = df["Eclipse_Type"].replace({
    "P": "Partial",
    "N": "Penumbral",
    "T": "Total",
    "T+": "Total",
    "T-": "Total",
    "Nx": "Penumbral",
    "Ne": "Penumbral",
    "Nb": "Penumbral"
})
```

```python
# replace P -> Partial.....

df["Eclipse_Type"]  = df["Eclipse_Type"].replace({
    "P": "Partial",
    "N": "Penumbral",
    "T": "Total",
    "T+": "Total (north)",
    "T-": "Total (south)",
    "Nx": "Penumbral (total)",
    "Ne": "Penumbral (end)",
    "Nb": "Penumbral (begin)"
})
```
